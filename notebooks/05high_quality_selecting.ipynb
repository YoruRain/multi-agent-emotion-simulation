```python
import pandas as pd

TOPIC_WEIBO_PATH = r"..\data\cleaned\topic_weibo.parquet"
TOPIC_COMMENT_PATH = r"..\data\cleaned\topic_comment.parquet"
USER_INFO_PATH = r"..\data\cleaned\user_info.parquet"
USER_WEIBO_PATH = r"..\data\cleaned\user_weibo.parquet"


df_topic_weibo = pd.read_parquet(TOPIC_WEIBO_PATH)
df_topic_comment = pd.read_parquet(TOPIC_COMMENT_PATH)
df_user_info = pd.read_parquet(USER_INFO_PATH)
df_user_weibo = pd.read_parquet(USER_WEIBO_PATH)
```

```python
import pandas as pd

TOPIC_WEIBO_SUMMARY_PATH = r"..\data\interim\topic_weibo_summary.parquet"

df_topic_weibo_summary = pd.read_parquet(TOPIC_WEIBO_SUMMARY_PATH)

df_hq_topic_weibo = df_topic_weibo_summary[df_topic_weibo_summary["topic_value"] == 3]

df_hq_topic_comment = df_topic_comment[df_topic_comment["weibo_id"].isin(df_hq_topic_weibo["weibo_id"])]

df_hq_user_info = df_user_info[df_user_info["user_id"].isin(df_hq_topic_comment["user_id"])]

df_hq_user_weibo = df_user_weibo[df_user_weibo["user_id"].isin(df_hq_user_info["user_id"])]

df_hq_user_weibo_2025 = df_hq_user_weibo[df_hq_user_weibo['year'] == 2025]
```

```python
df_hq_topic_weibo.to_parquet(r"..\data\high_quality\topic_weibo.parquet")
df_hq_topic_comment.to_parquet(r"..\data\high_quality\topic_comment.parquet")
df_hq_user_info.to_parquet(r"..\data\high_quality\user_info.parquet")
df_hq_user_weibo.to_parquet(r"..\data\high_quality\user_weibo.parquet")
df_hq_user_weibo_2025.to_parquet(r"..\data\high_quality\user_weibo_2025.parquet")
```
