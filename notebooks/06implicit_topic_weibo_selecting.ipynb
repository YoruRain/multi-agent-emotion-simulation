```python
import pandas as pd

df_weibo_topic = pd.read_parquet(r"..\data\profile\weibos\subject_profile\user_weibo_topic_signals.parquet")
```

```python
df_candidate = df_weibo_topic.loc[
    (df_weibo_topic["text_quality"] >= 3) &
    (df_weibo_topic["is_repost"] == False) & 
    (df_weibo_topic["explicit_topic_categories"].isna())
    , ["weibo_id", "user_id", "content", "text_quality", "is_repost", 
       "explicit_topic_categories", "signal_confidence"]
]
```

```python
df_candidate.to_parquet(r"..\data\interim\candidate_weibos_for_implicit_topic.parquet")
```
